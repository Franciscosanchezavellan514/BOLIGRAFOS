```python
import pandas as pd
import matplotlib.pyplot as plt

#grafico de análisis de Tiempo de Entrega
file_path = "./Pen Sales Data.xlsx"
df_pen_sales = pd.read_excel(file_path, sheet_name="Pen Sales")

print(df_pen_sales["Purchase Date"])
print(df_pen_sales["Delivery Date"])


df_pen_sales["tiempo de entrega"] = (df_pen_sales["Delivery Date"] - df_pen_sales["Purchase Date"]).dt.days


print(df_pen_sales.dtypes)
print(df_pen_sales["tiempo de entrega"])

tiempo_medio_entrega = df_pen_sales.groupby("Item")["tiempo de entrega"].mean().sort_values()
print(tiempo_medio_entrega)

plt.figure(figsize=(10, 5))
tiempo_medio_entrega.plot(kind="bar", color="orange")

plt.title("Tiempo promedio de entrega por producto")
plt.xlabel("Tipo de producto")
plt.ylabel("Días promedio de entrega")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
