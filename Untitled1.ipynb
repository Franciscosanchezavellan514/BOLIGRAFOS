```python
import pandas as pd
import matplotlib.pyplot as plt

# Ranking de popularidad de productos

file_path = "./Pen Sales Data.xlsx"
df_pen_sales = pd.read_excel(file_path, sheet_name="Pen Sales")
df_pen_sales["Item"] = df_pen_sales["Item"].str.strip()
df_product_ranking = df_pen_sales["Item"].value_counts()

print(df_pen_sales.dtypes)
print(df_product_ranking)

plt.figure(figsize=(10, 5))
df_product_ranking.plot(kind="barh", color="green")

plt.title("Ranking de popularidad de productos")
plt.xlabel("Cantidad de ventas")
plt.ylabel("Tipo de producto")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
